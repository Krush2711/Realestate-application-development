```python
import pandas as pd 
import numpy as np
```

```python
df = pd.read_csv('data_houses.csv')
```

```python
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.drop(['link', 'rating', 'property_id'], axis =1 , inplace=True)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.dropna(subset='price', inplace=True)
```

```python
def treat_price(x):
    if type(x) == float:
        return x
    else:
        if x[1] == 'Lac':
            return round(float(x[0])/100,2)
        else:
            return round(float(x[0]),2)
```

```python
df.head()
```

```python
df.price.value_counts()
```

```python
df= df[df["price"] != "Price on Request"]
```

```python
df['price'].value_counts()
```

```python
df['price'] = df['price'].str.split(' ').apply(treat_price)
```

```python
df.head()
```

```python
df['area'].value_counts()
```

```python
df['area']
```

```python
df['area'] = df['area'].str.split(')').str.get(0).str.split(' ').str.get(0).str.replace('(', '').astype(int)
```

```python
df['area']  = df['area'].apply(lambda x : x * 10.76)
```

```python
df.head()
```

```python
df.rename(columns={'rate':'price_per_sqft'},inplace=True)
df.head()
```

```python
df['bedRoom'].isnull().sum()
```

```python
df['bedRoom'] = df['bedRoom'].str.split(' ').str.get(0).astype(int)
```

```python
df['bathroom'] = df['bathroom'].str.split(' ').str.get(0).astype(int)
```

```python
df['balcony'] = df['balcony'].str.split(' ').str.get(0).str.replace('+', '').str.replace('No', '0').astype(int)
```

```python
df.head()
```

```python
df.info()
```

```python
df['noOfFloor'] = df['noOfFloor'].str.split(' ').str.get(0).str.replace('Ground', '0').str.replace('Basement', '0').str.replace('Lower', '0').str.extract(r'(\d+)').astype(float)
```

```python
df.info()
```

```python
df['society'] = (
    df['society']
    .str.replace(r'\d+(\.\d+)?\s?★', '', regex=True)
    .str.strip()
    .str.lower())
df['society'].value_counts().shape
```

```python
df['price_per_sqft'] = df['price_per_sqft'].str.split('/').str.get(0).str.split(' ').str.get(1).str.replace(',', '').astype(int)
```

```python
df.info()
```

```python
df.rename(columns={'noOfFloor':'floorNum'},inplace=True)
df.head()
```

```python
df.info()
```

```python
df['facing'].fillna('DK', inplace=True)
```
